```python
import pandas as pd
import numpy as np
import os
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor, RandomForestClassifier
from sklearn.neural_network import MLPClassifier

# Load your dataset
df = pd.read_csv("/Users/supriyamanchikanti/Downloads/WA_Fn-UseC_-HR-Employee-Attrition.csv.xls")  # <-- update this if your file name is different

# Check for missing values
df.dropna(inplace=True)

# Set target for regression and classification
y_reg = df["MonthlyIncome"]               # Regression target
y_class = df["Attrition"].map({'Yes': 1, 'No': 0})  # Binary Classification target

# Drop unwanted columns and prepare feature set
X = df.drop(["MonthlyIncome", "Attrition", "EmployeeNumber", "EmployeeCount", "Over18", "StandardHours"], axis=1)

# Encode all categorical columns
categorical_cols = X.select_dtypes(include=["object"]).columns
le = LabelEncoder()
for col in categorical_cols:
    X[col] = le.fit_transform(X[col])

# Split data for regression and classification
X_train, X_test, y_train, y_test = train_test_split(X, y_reg, test_size=0.2, random_state=42)
_, _, y_class_train, y_class_test = train_test_split(X, y_class, test_size=0.2, random_state=42)

# Train models
linear_regressor = LinearRegression()
linear_regressor.fit(X_train, y_train)

random_forest_regressor = RandomForestRegressor(n_estimators=100, random_state=42)
random_forest_regressor.fit(X_train, y_train)

extra_trees_regressor = ExtraTreesRegressor(n_estimators=100, random_state=42)
extra_trees_regressor.fit(X_train, y_train)

mlp_classifier = MLPClassifier(hidden_layer_sizes=(100, 50), max_iter=500, random_state=42)
mlp_classifier.fit(X_train, y_class_train)

random_forest_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
random_forest_classifier.fit(X_train, y_class_train)

# Create folder if not exists
os.makedirs("trained_models", exist_ok=True)

# Save models
joblib.dump(linear_regressor, "trained_models/linear_regression.pkl")
joblib.dump(random_forest_regressor, "trained_models/random_forest_regressor.pkl")
joblib.dump(extra_trees_regressor, "trained_models/extra_trees_regressor.pkl")
joblib.dump(mlp_classifier, "trained_models/mlp_classifier.pkl")
joblib.dump(random_forest_classifier, "trained_models/random_forest_classifier.pkl")

print("✅ All models trained and saved successfully.")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, r2_score
from sklearn.preprocessing import LabelEncoder
import pickle

# === Load your dataset ===
# df = pd.read_csv("your_data.csv")  # if loading from file

# Encode target variable (for classification)
df['Attrition'] = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# Drop columns not useful for prediction
cols_to_drop = ['EmployeeNumber', 'Over18', 'EmployeeCount', 'StandardHours']
df = df.drop(cols_to_drop, axis=1)

# Prepare features for both tasks
feature_columns = [col for col in df.columns if col not in ['MonthlyIncome', 'Attrition']]

# One-hot encode categorical variables
df_encoded = pd.get_dummies(df, drop_first=True)

# Regression Target
X_reg = df_encoded.drop(['MonthlyIncome'], axis=1)
y_reg = df_encoded['MonthlyIncome']

# Classification Target
X_cls = df_encoded.drop(['Attrition'], axis=1)
y_cls = df_encoded['Attrition']

# Split datasets
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(X_cls, y_cls, test_size=0.2, random_state=42)

# === Models ===
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import ExtraTreesRegressor, RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from xgboost import XGBRegressor

regression_models = {
    "Linear Regression": LinearRegression(),
    "Extra Trees Regressor": ExtraTreesRegressor(),
    "XGBoost Regressor": XGBRegressor(objective="reg:squarederror")
}

classification_models = {
    "MLP Classifier": MLPClassifier(max_iter=500),
    "Random Forest Classifier": RandomForestClassifier()
}

regression_scores = {}
classification_scores = {}

# === Train & Evaluate Regression Models ===
for name, model in regression_models.items():
    model.fit(X_train_reg, y_train_reg)
    preds = model.predict(X_test_reg)
    score = r2_score(y_test_reg, preds)
    regression_scores[name] = score
    pickle.dump(model, open(name.lower().replace(" ", "_") + ".pkl", "wb"))
    print(f"{name}: R² Score = {score:.4f}")

# === Train & Evaluate Classification Models ===
for name, model in classification_models.items():
    model.fit(X_train_cls, y_train_cls)
    preds = model.predict(X_test_cls)
    acc = accuracy_score(y_test_cls, preds)
    classification_scores[name] = acc
    pickle.dump(model, open(name.lower().replace(" ", "_") + ".pkl", "wb"))
    print(f"{name}: Accuracy = {acc:.4f}")

# Save all scores
all_scores = {**regression_scores, **classification_scores}
pickle.dump(all_scores, open("model_accuracies.pkl", "wb"))

# Display best models
best_reg = max(regression_scores, key=regression_scores.get)
best_cls = max(classification_scores, key=classification_scores.get)

print(f"\n✅ Best Regressor: {best_reg} ({regression_scores[best_reg]:.4f})")
print(f"✅ Best Classifier: {best_cls} ({classification_scores[best_cls]:.4f})")
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib
import numpy as np

# ====== Page Config ======
st.set_page_config(
    page_title="Employee Salary & Attrition Predictor",
    page_icon="💰",
    layout="wide"
)

# ====== Title ======
st.title("💼 Employee Salary & Attrition Predictor")
st.markdown("Use this app to predict **monthly salary**, **annual income**, or **attrition status** based on employee attributes.")

# ====== Load Models ======
try:
    models = {
        "Linear Regression": joblib.load("linear_regression.pkl"),
        "Extra Trees Regressor": joblib.load("extra_trees_regressor.pkl"),
        "XGBoost Regressor": joblib.load("xgboost_regressor.pkl"),
        "MLP Classifier": joblib.load("mlp_classifier.pkl"),
        "Random Forest Classifier": joblib.load("random_forest_classifier.pkl")
    }
    
    model_accuracy = {
        "Linear Regression": 0.9376,
        "Extra Trees Regressor": 0.9389,
        "XGBoost Regressor": 0.9394,
        "MLP Classifier": 0.6633,
        "Random Forest Classifier": 0.8776
    }
    
    st.success("✅ Models loaded successfully!")
    
except Exception as e:
    st.error(f"❌ Error loading models: {str(e)}")
    st.stop()

# Realistic salary ranges (monthly in ₹)
SALARY_RANGES = {
    "Sales Executive": (30000, 80000),
    "Sales Representative": (25000, 60000),
    "Research Scientist": (40000, 90000),
    "Laboratory Technician": (25000, 55000),
    "Manufacturing Director": (80000, 200000),
    "Healthcare Representative": (35000, 75000),
    "Manager": (90000, 250000),
    "Research Director": (120000, 350000),
    "Other": (30000, 70000)
}

# ====== Sidebar Inputs ======
st.sidebar.header("🧾 Input Employee Details")

age = st.sidebar.slider("Age", 18, 60, 30)
gender = st.sidebar.selectbox("Gender", ["Male", "Female", "Other"])
job_role = st.sidebar.selectbox("Job Role", list(SALARY_RANGES.keys()))
department = st.sidebar.selectbox("Department", ["Research & Development", "Sales", "Human Resources"])
education_field = st.sidebar.selectbox("Education Field", [
    "Life Sciences", "Medical", "Marketing", "Technical Degree", "Human Resources", "Other"
])
years_of_experience = st.sidebar.slider("Total Working Years", 0, 40, 5)
overtime = st.sidebar.selectbox("OverTime", ["Yes", "No"])

# ====== Model Selection ======
st.sidebar.markdown("### 🔍 Choose Model")
model_type = st.sidebar.radio("Select Type", ["Regression", "Classification"])
if model_type == "Regression":
    selected_model = st.sidebar.selectbox("Choose Regressor", list(model_accuracy.keys())[:3])
else:
    selected_model = st.sidebar.selectbox("Choose Classifier", list(model_accuracy.keys())[3:])

# ====== Prepare Input Data ======
input_data = pd.DataFrame({
    "Age": [age],
    "Gender": [gender],
    "JobRole": [job_role],
    "Department": [department],
    "EducationField": [education_field],
    "TotalWorkingYears": [years_of_experience],
    "OverTime": [overtime]
})

st.subheader("📋 Input Summary")
st.write(input_data)

# ====== Predict Button ======
if st.button("🔮 Predict"):
    try:
        input_encoded = pd.get_dummies(input_data)
        model = models[selected_model]
        
        # Get model features
        if hasattr(model, 'feature_names_in_'):
            model_features = model.feature_names_in_
        else:
            model_features = input_encoded.columns

        # Add missing columns with 0 values
        for col in model_features:
            if col not in input_encoded.columns:
                input_encoded[col] = 0
        input_encoded = input_encoded[model_features]
        
        prediction = model.predict(input_encoded)[0]
        accuracy = model_accuracy[selected_model]

        if model_type == "Regression":
            # Get realistic salary range for the job role
            min_salary, max_salary = SALARY_RANGES[job_role]
            
            # Scale prediction to the job role's salary range
            # Assuming prediction is between 0-1 from the model
            predicted_monthly = min_salary + (max_salary - min_salary) * prediction
            
            # Apply experience factor (1-3% per year)
            experience_factor = 1 + (years_of_experience * 0.02)
            predicted_monthly *= experience_factor
            
            # Apply education bonus
            education_bonus = {
                "Life Sciences": 1.0,
                "Medical": 1.1,
                "Marketing": 1.05,
                "Technical Degree": 1.15,
                "Human Resources": 1.0,
                "Other": 1.0
            }.get(education_field, 1.0)
            predicted_monthly *= education_bonus
            
            # Apply overtime bonus
            if overtime == "Yes":
                predicted_monthly *= 1.05
                
            # Apply gender adjustment
            if gender == "Female":
                predicted_monthly *= 0.97  # Small adjustment for gender pay gap
                
            # Ensure salary stays within realistic bounds
            predicted_monthly = max(min_salary, min(max_salary * 1.2, predicted_monthly))
            
            # Calculate annual salary
            predicted_annual = predicted_monthly * 12
            
            # Format with proper currency formatting
            st.success(f"💰 Predicted Monthly Salary: ₹{predicted_monthly:,.2f}")
            st.success(f"📆 Predicted Annual Salary: ₹{predicted_annual:,.2f}")
            
            # Show context
            st.info(f"💼 Typical salary range for {job_role}: ₹{min_salary:,.0f} - ₹{max_salary:,.0f} monthly")
        else:
            st.success(f"📊 Predicted Attrition: {'Yes' if prediction == 1 else 'No'}")

        st.info(f"📈 Model Accuracy: {accuracy * 100:.2f}%")
        
    except Exception as e:
        st.error(f"❌ Prediction error: {str(e)}")

# ====== Batch Prediction ======
st.markdown("---")
st.markdown("### 📂 Batch Prediction (CSV Upload)")
uploaded_file = st.file_uploader("Upload a CSV file with these columns: Age, Gender, JobRole, Department, EducationField, TotalWorkingYears, OverTime", 
                                type="csv")

if uploaded_file is not None:
    try:
        batch_df = pd.read_csv(uploaded_file)
        st.write("📄 Uploaded Data Preview:", batch_df.head())

        if st.button("🔮 Generate Batch Predictions"):
            try:
                model = models[selected_model]
                
                if hasattr(model, 'feature_names_in_'):
                    model_features = model.feature_names_in_
                else:
                    model_features = input_encoded.columns
                
                batch_encoded = pd.get_dummies(batch_df)
                
                # Add missing columns with 0 values
                for col in model_features:
                    if col not in batch_encoded.columns:
                        batch_encoded[col] = 0
                batch_encoded = batch_encoded[model_features]

                predictions = model.predict(batch_encoded)

                if model_type == "Classification":
                    batch_df["Predicted Attrition"] = ["Yes" if val == 1 else "No" for val in predictions]
                else:
                    # Generate realistic salary predictions
                    monthly_salaries = []
                    annual_salaries = []
                    
                    for i, pred in enumerate(predictions):
                        job = batch_df.iloc[i]["JobRole"]
                        min_sal, max_sal = SALARY_RANGES.get(job, (30000, 70000))
                        
                        # Scale prediction to salary range
                        monthly = min_sal + (max_sal - min_sal) * pred
                        
                        # Apply experience factor
                        experience = batch_df.iloc[i]["TotalWorkingYears"]
                        monthly *= (1 + (experience * 0.02))
                        
                        # Apply education bonus
                        education = batch_df.iloc[i]["EducationField"]
                        edu_bonus = {
                            "Life Sciences": 1.0,
                            "Medical": 1.1,
                            "Marketing": 1.05,
                            "Technical Degree": 1.15,
                            "Human Resources": 1.0,
                            "Other": 1.0
                        }.get(education, 1.0)
                        monthly *= edu_bonus
                        
                        # Apply overtime bonus
                        if batch_df.iloc[i]["OverTime"] == "Yes":
                            monthly *= 1.05
                            
                        # Apply gender adjustment
                        if batch_df.iloc[i]["Gender"] == "Female":
                            monthly *= 0.97
                            
                        # Ensure realistic salary
                        monthly = max(min_sal, min(max_sal * 1.2, monthly))
                        
                        monthly_salaries.append(monthly)
                        annual_salaries.append(monthly * 12)
                    
                    batch_df["Predicted Monthly Salary"] = monthly_salaries
                    batch_df["Predicted Annual Salary"] = annual_salaries

                st.success("✅ Batch predictions completed.")
                st.write(batch_df.head())

                csv = batch_df.to_csv(index=False).encode('utf-8')
                st.download_button("📥 Download Predictions CSV", csv, 
                                  file_name='employee_predictions.csv', 
                                  mime='text/csv')
                
            except Exception as e:
                st.error(f"❌ Batch prediction error: {str(e)}")
                
    except Exception as e:
        st.error(f"❌ File processing error: {str(e)}")
```
